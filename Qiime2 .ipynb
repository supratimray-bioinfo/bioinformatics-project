```python
qiime tools import \
    --type 'SampleData[PairedEndSequencesWithQuality]' \
    --input-path q2_manifest.csv \
    --output-path 01_import/pairend_sequences.qza \
    --input-format PairedEndFastqManifestPhred33
```

```python
qiime demux summarize \
   --i-data 01_import/pairend_sequences.qza \
   --o-visualization 01_import/pairend_sequences.qzv;
```

```python
ls
```

```python
mkdir 02_denoise  
qiime dada2 denoise-paired \
  --i-demultiplexed-seqs 01_import/pairend_sequences.qza \
  --p-trim-left-f 0 \
  --p-trim-left-r 0 \
  --p-trunc-len-f 150 \
  --p-trunc-len-r 140 \
  --p-n-threads 6 \
  --o-table 02_denoise/dada2_table.qza \
  --o-representative-sequences 02_denoise/rep_seqs.qza \
  --o-denoising-stats 02_denoise/denoising_stats.qza;
```

```python
qiime feature-table tabulate-seqs \
  --i-data 02_denoise/rep_seqs.qza \
  --o-visualization 02_denoise/rep_seqs.qzv;
 
qiime metadata tabulate \
  --m-input-file 02_denoise/denoising_stats.qza \
  --o-visualization 02_denoise/denoising_stats.qzv;
```

```python
# downloading the naive bayes classifier for silva v4 region
wget https://data.qiime2.org/2021.8/common/silva-138-99-515-806-nb-classifier.qza;
```

```python
qiime feature-classifier classify-sklearn \
  --i-reads rep_seqs.qza \
  --i-classifier silva-138-99-515-806-nb-classifier.qza \
  --p-n-jobs 20 \
  --o-classification taxonomy.qza;
```

```python
# summarize tables at different levels
mkdir 05_summary
qiime taxa collapse \
  --i-table dada2_table.qza \
  --i-taxonomy taxonomy.qza \
  --p-level 2 \
  --o-collapsed-table 05_summary/table-l2.qza;
  
qiime taxa collapse \
  --i-table dada2_table.qza \
  --i-taxonomy taxonomy.qza \
  --p-level 3 \
  --o-collapsed-table 05_summary/table-l3.qza;

qiime taxa collapse \
  --i-table dada2_table.qza \
  --i-taxonomy taxonomy.qza \
  --p-level 4 \
  --o-collapsed-table 05_summary/table-l4.qza;
  
qiime taxa collapse \
  --i-table dada2_table.qza \
  --i-taxonomy taxonomy.qza \
  --p-level 5 \
  --o-collapsed-table 05_summary/table-l5.qza; 
  
qiime taxa collapse \
  --i-table dada2_table.qza \
  --i-taxonomy taxonomy.qza \
  --p-level 6 \
  --o-collapsed-table 05_summary/table-l6.qza;
  
qiime taxa collapse \
  --i-table dada2_table.qza \
  --i-taxonomy taxonomy.qza \
  --p-level 7 \
  --o-collapsed-table 05_summary/table-l7.qza
```
